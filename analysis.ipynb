```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data=pd.read_csv("dataset/dataset.csv")
data.head(10)
```

```python
data.describe()
```

```python
plt.scatter(data['Age'],data['LAD'])
```

```python
heatmap_features=['DM',"HTN","Current Smoker","EX-Smoker","FH"]
for i in heatmap_features:
    plt.figure(figsize=(10,6))
    sns.heatmap(data[[i,'LAD']], annot=True)
    plt.title(f'Correlation between {i} and LAD')
    plt.show()
```

```python
predict_columns=['LAD','LCX','RCA','Cath']
for i in data.columns:
    if i not in predict_columns:
        plt.scatter(data[i],data['LAD'])

        plt.xlabel(i)
        plt.ylabel('LAD')
        plt.show()
```

```python
correlation_matrix = data.corr()
print(correlation_matrix)
```
